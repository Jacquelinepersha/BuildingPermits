```python
from faker import Faker
import pandas as pd
```

```python
fake = Faker()
data = []
for _ in range(50):
    data.append({
        "Name": fake.name(),
        "Email": fake.email(),
        "Product": fake.word().title(),
        "Value": round(fake.random_number(digits=3), 2),
        "LastContact": fake.date_this_year(),
        "Status": "Abandoned"
    })

df = pd.DataFrame(data)
df.to_csv("sample_leads.csv", index=False)
```
